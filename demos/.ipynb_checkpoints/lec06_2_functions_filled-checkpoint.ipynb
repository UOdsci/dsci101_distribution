## Defining Functions ##

```python
## Define a function named spread:
## - input: values (an array of numbers)
## - output: the difference between the max and min of values (a number)
def spread(values):
    return max(values) - min(values)
```

```python
spread(make_array(1, 10, 60000, 99, 78))
```

```python
## Let's write a function that takes an input and multiplies it by 3
def triple(x):
    return x*3
```

```python
triple(3)
```

```python
num = 4
triple(num)
```

```python
triple(num * 5)
```

```python
## You don't need to define the input type "Type Agnostic"
```

```python
triple("ha")
```

```python
triple(np.arange(9))
```

### Functions can take multiple arguments

$ h^2 = x^2 + y^2 \hspace{20 pt} => \hspace{20 pt} h = \sqrt{ x^2 + y^2 } $

```python
def hypotenuse(x,y):
    hypot_squared = x**2 + y**2
    return hypot_squared ** 0.5
```

```python
hypotenuse(9, 12)
```

```python
hypotenuse(2, 2)
```
