# Control statements: if

```python
num = 42
if num > 100:
    print('greater')
else:
    print('not greater')
print('done')

# here's the if else statement from the slides
# let's see what happens as we change the value of num
```

```python
# Conditional statements don’t have to include an else.
# If there isn’t one, Python simply does nothing if the test is false
```

```python
num = 107
print('before conditional...')
if num > 100:
    print(num,' is greater than 100')
print('...after conditional')

# again, we can see how it changes with the value of num
# or with changing the conditional
```

```python
# We can also chain several tests together using elif, 
# which is short for “else if”. 
# The following Python code uses elif to print the sign of a number.
```

```python
num = 42

if num > 0:
    print(num, 'is positive')
elif num == 0:
    print(num, 'is zero')
else:
    print(num, 'is negative')
print('all done')

# and again, let's try it out and see what happens when we change things
# what will be printed if num is -1.5?
```

```python
 # We can also combine tests using and and or. 
 # and is only true if both parts are true:
```

```python
if (1 > 0) and (-1 >= 0):
    print('both parts are true')
else:
    print('at least one part is false')

# how can we change the components of the conditional to make the whole thing true?
```

```python
# while or is true if at least one part is true:
```

```python
if (1 < 0) or (1 >= 0):
    print('at least one test is true')

# how can we change the components of the conditional to make the whole thing false?
```
